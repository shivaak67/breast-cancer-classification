```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns

PROJECT_ROOT = Path.cwd().parent if (Path.cwd().parent / "src").exists() else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / "src"))

from preprocess import preprocess_data
from train import train_model, save_artifacts
from evaluate import evaluate_model

data = preprocess_data()

X_train_scaled = data["X_train_scaled"]
X_test_scaled = data["X_test_scaled"]
y_train = data["y_train"]
y_test = data["y_test"]
target_names = data["target_names"]
```

```python
model = train_model(X_train_scaled, y_train)
save_artifacts(model, data["scaler"])
model
```

```python
results = evaluate_model(model, X_test_scaled, y_test, target_names)
y_pred = results["y_pred"]
```

```python
results["accuracy"]
```

```python
cm = results["confusion_matrix"]
cm
```

```python
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["malignant", "benign"],
    yticklabels=["malignant", "benign"],
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

```python
print(results["classification_report"])
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train_scaled, y_train)

rf_pred = rf_model.predict(X_test_scaled)
accuracy_score(y_test, rf_pred)
```
